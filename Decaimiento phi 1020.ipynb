```
mu=\[Mu]
nu=\[Nu]
a=\[Alpha]
b=\[Beta]
n=\[Eta]

SF[p_]:=I FAD[{p,m}]
DF[mu_,nu_,k_]:=-FAD[k] I MTD[mu,nu]
SF[p]
DF[a,b,k]

SPD[p,p]=x
SPD[p,q]=w
SPD[q,q]=y
```

## $p$ - momento de la partícula escalar y $k$ - el momento a integrar

```
M=e*(2*FVD[p,mu]-FVD[k,mu])*e*(2*FVD[p,nu]-FVD[k,nu])*DF[mu,nu,k]*SF[p-k]
```

## Integrando $M$ con respecto de k

```
resultInt1=TID[M,k,ToPaVe -> True, Prefactor -> 1/(2Pi)^D];
Sigma=FCReplaceD[resultInt1, D -> 4]
Print["The result is given in terms of Passarino-Veltman"]
```

## Derivada de $\Sigma$ con respecto de $p^2$. (Se nombra a $p^2=x$)

```
der=D[Sigma,x]
```

## La derivada de la amplitud de dispersión evaluada en $p^2 = m^2$ es

```
der/.x->m^2
```

## Ahora, la corrección de vértice (interacción de Coulomb), tiene la forma:

## Aqui $MV$ es la aportacion del vertice en la matriz de disperson, $p$ es el momento de las partículas, al ser la misma particula pero con cargas contrarias, se usa la misma variable

```
MV=I*g*(FVD[p,a]-FVD[q,a]+2*FVD[k,a])*FVD[n,a]*(SF[p+k])*(-I*e*(2*FVD[p,mu]+FVD[k,mu])*(DF[mu,nu,k])*I*e*(2*FVD[q,nu]-FVD[k,nu])*SF[q-k])
```

## Integrando $MV$ con respecto de k

```
resultInt2=TID[MV,k,ToPaVe -> True, Prefactor -> 1/(2Pi)^D];
SigmaV=FCReplaceD[resultInt2, D -> 4]
Print["The result is given in terms of Passarino-Veltman"]
```

## Evaluando este resultado en $p^2=m^2$

```
SVx=SigmaV/.x->m^2
```

```
SVy=SVx/.y->m^2
```

## El resultado ya simplifado es

```
SVy //Simplify
```
